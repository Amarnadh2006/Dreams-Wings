```python

import numpy as np
x=np.array(dataset["YearsExperience"])
y=np.array(dataset["Salary"])
print(x)
print(y)
```

```python
x=x.reshape(-1,1)
y=y.reshape(-1,1)
print(x)
print(y)
```

```python
x_norm=(x-x.mean())/x.std()
y_norm=(y-y.mean())/y.std()
print(x_norm)
print(y_norm)
```

#TASK 3

```python
import numpy as np

m = 0.0 #slope
b = 0.0 #intercept

def predict(x, m, b):
    return m * x + b

def mean_square_error(X, Y, m, b):
    y_pred = predict(X, m, b)
    n = len(X)
    return (1/n) * np.sum((Y - y_pred)**2)

# Calculate cost
cost = mean_square_error(x_norm, y_norm, m, b)

print("slope (m):", m)
print("intercept (b):", b)
print("cost (MSE):", cost)
```

#TASK 4


```python
learning_rate=0.01
epochs=1000
for epochs in range(epochs):
    Y_pred=predict(x_norm,m,b)
    n=len(x_norm)
    #gradients
    dm=(-2/n)*np.sum(x_norm*(y_norm-Y_pred))
    db=(-2/n)*np.sum(y_norm-Y_pred)
    #update parameters
    m=m-learning_rate*dm
    b=b-learning_rate*db
    cost=mean_square_error(x_norm,y_norm,m,b)
    if epochs%100==0:
        print("epoch:",epochs,"cost:",cost)
#final slope,intercept and cost
print("final slope:",m)
print("final intercept:",b)
print("final cost:",cost)
```

#TASK 5

```python
#salary predict
Y_pred=predict(x_norm,m,b)
print(Y_pred)
#Mean squared error
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_norm, Y_pred)
print(f"Mean Squared Error: {mse:.2f}")
#plot
import matplotlib.pyplot as plt
plt.scatter(x_norm, y_norm, color='blue', label='Actual Data Points')
plt.plot(x_norm, Y_pred, color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary Prediction: Regression Line vs. Actual Data')
plt.legend()
plt.show()

```
